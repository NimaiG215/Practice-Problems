```python
#To accept 20 values form the user and save the list
num=[]
print("Please Enter 20 Numbers")
for i in range(20):
    val=int(input("Enter number "))
    num.append(val)
print("\nYour List:",num)

#a Count Similar Elements
element_map={}
for index, val in enumerate(num):
    if val not in element_map:
        element_map[val]=[]
        element_map[val].append(index)

#b  Count Even Odd values
even=0
odd=0
for val in num:
    if val%2==0:
        even+=1
    else:
        odd+=1
print("Even Count",even)
print("Odd Count",odd)
                        
                        

```

```python
# Accept 10 values from the user and save it ina list
num=[]
print("Please Enter 10 Numbers")
for i in range(10):
    val=int(input("Enter number "))
    num.append(val)
print("You Entered the following numbers:",num)

#Print in ascending order
num.sort()
print("ASCENDING ORDER:",num)

#Print in Descending Order
num.sort(reverse=True)
print("DESCENDING ORDER:",num)
```

```python
# To Accept two lists from the user and merging it
list1=[]
for i in range(5):
    val=int(input("Enter Value"))
    list1.append(val)
list2=[]
for i in range(5):
    val=int(input("Enter Value"))
    list2.append(val)

mergedlist=list1+list2
print("List 1 is:",list1)
print("List 2 is:",list2)
print("Merged List is:",mergedlist)
```

```python
#print the abbreviation of the month,given its number
months=["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sept","Oct","Nov","Dec"]
month_number=int(input("Enter month number(1-12):"))
if month_number>=1 and month_number<=12:
    print("Month abbreviation:",months[month_number-1])
else:
    print("Inavalid month number")
```
