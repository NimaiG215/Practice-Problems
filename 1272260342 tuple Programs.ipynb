```python
#to print 4th element form first amd last
num=(10,20,30,40,50,60,70,80,90)
print("Tuple:",num)
print("4th element from first:",num[3])
print("4th element from last:",num[-4])
```

```python
#to check wheter the element exists in a tuple
num=(10,20,30,40,50)
element=int(input("Enter element to search:"))
if element in num:
    print("Element exists in tuple")
else:
    print("Element does not exist in tuple")
```

```python
#converting list to tuple
num_list=[10,20,30,40,50]
num_tuple=tuple(num_list)
print(num_list)
print(num_tuple)
```

```python
#to find index of an element
num=(10,20,30,40,50)
element=int(input("Enter element"))
if element in num:
    print("Index of element:",num.index(element))
else:
    print("not found")
```

```python
#to relace last value of tuples in a list to 100
num=[(10,20,30),(40,50,60),(70,80,90)]
new_list=[]
for tup in num:
    new_tup=tup[:-1] + (100,)
    new_list.append(new_tup)
    print("Original List:",num)
    print("Updated List:",new_list)
```

```python
#program to swap first and last element of a tuple
num=(10,20,30,40,50)
new_tuple=(num[-1],) + num[1:-1] + (num[0],)
print("Original Tuple:",num)
print("Tuple after swapping:",new_tuple)
```
